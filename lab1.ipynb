```python
from openai import OpenAI

openai = OpenAI()
```

```python
messages = [{"role": "user", "content": "Tell me a fun fact"}]
response = openai.chat.completions.create(model="gpt-5.4-nano", messages=messages)
print(response.choices[0].message.content)
```

```python
question_prompt = (
    "Please propose a hard, challenging question to assess someone's IQ. "
    "Respond only with the question."
)
messages = [{"role": "user", "content": question_prompt}]
```

```python
response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
question = response.choices[0].message.content
print(question)
```

```python
messages = [{"role": "user", "content": question}]
response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
answer = response.choices[0].message.content
print(answer)
```

```python
from IPython.display import Markdown, display

display(Markdown(answer))
```

```python
message = f"""
Here is a question:
{question}

And here is a possible answer that might be correct or incorrect:
{answer}

Please evaluate if the answer is correct or incorrect.
"""

messages = [{"role": "user", "content": message}]
response = openai.chat.completions.create(model="gpt-5.4", messages=messages)
print(response.choices[0].message.content)
```

## Exercise

Chain three LLM calls:
1. Pick a business area worth exploring for Agentic AI
2. Identify a pain point in that industry
3. Propose an Agentic AI solution

```python
# Call 1 — pick a business area
messages = [{"role": "user", "content": "Pick a business area worth exploring for Agentic AI. Reply with just the area."}]
response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
business_area = response.choices[0].message.content
print(business_area)
```

```python
# Call 2 — pain point in that industry
messages = [{"role": "user", "content": f"In {business_area}, describe one challenging pain point ripe for an Agentic AI solution."}]
response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
pain_point = response.choices[0].message.content
print(pain_point)
```

```python
# Call 3 — propose a solution
messages = [{"role": "user", "content": f"Business area: {business_area}\nPain point: {pain_point}\n\nPropose an Agentic AI solution."}]
response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
print(response.choices[0].message.content)
```
